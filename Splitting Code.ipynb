# Globals


```python
data_dir='/content/drive/Shareddrives/Thesis_Preparation/THT_Medicinal/Augmented_data'
save_path='/content/drive/Shareddrives/Thesis_Preparation/THT_Medicinal'
save_path=create_dir(save_path,'data')

train_save=create_dir(save_path,'train')
test_save=create_dir(save_path,'test')
eval_save=create_dir(save_path,'eval')
```

# Test Train  Split

```python
labels= os.listdir(data_dir)
for label in labels:
    create_dir(train_save,label)
    create_dir(test_save,label)
    create_dir(eval_save,label)
```

```python
import os
import cv2
from tqdm import tqdm
from glob import glob

img_dim = 224
count = 0
train_ratio = 0.7
val_ratio = 0.15

for label in labels:
    img_paths = [img_path for img_path in tqdm(glob(os.path.join(data_dir, label, "*.*")))]

    nb_train = int(len(img_paths) * train_ratio)
    nb_val = int(len(img_paths) * val_ratio)

    train_paths = img_paths[:nb_train]
    val_paths = img_paths[nb_train:nb_train + nb_val]
    test_paths = img_paths[nb_train + nb_val:]

    # Process training images
    for img_path in tqdm(train_paths):
        img = cv2.imread(img_path)
        img = cv2.resize(img, (img_dim, img_dim))
        cv2.imwrite(os.path.join(train_save, label, f"{count}.png"), img)
        count += 1

    # Process validation images
    for img_path in tqdm(val_paths):
        img = cv2.imread(img_path)
        img = cv2.resize(img, (img_dim, img_dim))
        cv2.imwrite(os.path.join(eval_save, label, f"{count}.png"), img)
        count += 1

    # Process test images
    for img_path in tqdm(test_paths):
        img = cv2.imread(img_path)
        img = cv2.resize(img, (img_dim, img_dim))
        cv2.imwrite(os.path.join(test_save, label, f"{count}.png"), img)
        count += 1
```
