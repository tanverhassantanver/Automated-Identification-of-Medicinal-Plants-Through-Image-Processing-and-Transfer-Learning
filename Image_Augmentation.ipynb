```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import shutil
```

```python
# Paths to your dataset and where to save the augmented images in Google Drive
data_dir = "/content/drive/Shareddrives/Thesis_Preparation/THT_Medicinal/Medicinal_Leaf"
augmented_data_dir = "/content/drive/Shareddrives/Thesis_Preparation/THT_Medicinal/Augmented_data"
```

```python
# Define classes
classes = ['Aristolochia_indica', 'Centella_asiatica', 'Giant_calotrope', 'Green_chiretta', 'Terminalia_bellirica']

# Function to create directories if they don't exist
def create_dir(directory):
    if not os.path.exists(directory):
        os.makedirs(directory)

# Set up image data generator with augmentations
datagen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
```

```python
# Function to augment and save images
def augment_and_save(class_name, num_augmented_images):
    class_dir = os.path.join(data_dir, class_name)
    save_dir = os.path.join(augmented_data_dir, class_name)
    create_dir(save_dir)  # Create class directory for augmented data

    # Loop through images in each class
    for img_name in os.listdir(class_dir):
        if img_name.endswith(('.jpg', '.jpeg', '.png')):
            img_path = os.path.join(class_dir, img_name)
            img = tf.keras.preprocessing.image.load_img(img_path, target_size=(224, 224))
            x = tf.keras.preprocessing.image.img_to_array(img)
            x = np.expand_dims(x, axis=0)

            # Create a new batch of augmented images
            aug_iter = datagen.flow(x, batch_size=1)

            # Generate and save images
            for i in range(num_augmented_images):
                aug_img = next(aug_iter)[0].astype('uint8')
                aug_img_name = f"{os.path.splitext(img_name)[0]}_aug_{i}.png"
                save_path = os.path.join(save_dir, aug_img_name)
                tf.keras.preprocessing.image.save_img(save_path, aug_img)
```

```python
# Specify the number of augmented images per original image
num_augmented_images_per_image = 5

# Generate and save augmented images for each class
for class_name in classes:
    augment_and_save(class_name, num_augmented_images=num_augmented_images_per_image)

print("Augmentation complete. Augmented data saved to Google Drive.")
```
